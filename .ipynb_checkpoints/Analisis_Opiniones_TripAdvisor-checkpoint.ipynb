## 2. Nubes de palabras:
Primero se crean las bolsas de términos para cada uno de los lugares y los guardo en un diccionario, para ello se usará el pesado tfidf y se normaliza con l2:

```python
# Se eliminan las stopwords, para obtener mejor informacion:
from nltk.tokenize import word_tokenize
nltk.download('stopwords')
nltk.download('punkt')
stop_words_esp = set(stopwords.words('spanish'))
stop_words_eng = set(stopwords.words('english'))
for lugar in Lugares:
    lugar['Opinión'] = lugar['Opinión'].apply(lambda text: ' '.join([word for word in word_tokenize(text) if word.lower() not in stop_words_esp and word.lower() not in stop_words_eng]))
```

```python
# Sea crean diccionarios que contienen las bolsas de palabras por lugar, y las features que se usan
vectorizer = TfidfVectorizer(norm='l2')
i = 0
BoW = dict()
Caracteristicas = dict()
for lugar in Lugares:
    BoW[Lugares_str[i]] = vectorizer.fit_transform(lugar['Opinión'])
    Caracteristicas[Lugares_str[i]] = vectorizer.get_feature_names_out()
    i += 1
```

```python
# Se usa chi2 para seleccionar las mejores caracteristicas
i = 0
feats = SelectKBest(chi2, k=50)
for lugar in Lugares:
    print("Top 50 palabras más relevantes - ", Lugares_str[i])
    feats.fit(BoW[Lugares_str[i]], lugar['Escala'])
    best = feats.get_support(indices=True)
    scores = feats.scores_
    sorted_scores_indices = sorted(enumerate(scores), key=lambda x: x[1], reverse=True)
    top_k_features = sorted_scores_indices[:50]
    w = {Caracteristicas[Lugares_str[i]][k]: valor for k, valor in top_k_features}
    wordcloud = WordCloud(width=800, height=400).generate_from_frequencies(w)
    plt.imshow(wordcloud, interpolation='bilinear')
    plt.axis('off')
    plt.show()
    i += 1
```

## 3. Descubrimiento automático de tópicos con LSA.

```python
# Se usa LSA para obtener los topicos mas relevantes para cada lugar y para cada uno de los subgrupos pedidos: 
from sklearn.decomposition import TruncatedSVD
i = 0
for lugar in Lugares:
    print("-"*90)
    print("Tópicos más relevantes en el subgrupo de hombres - ", Lugares_str[i])
    lsa_model = TruncatedSVD(n_components=3)
    indices = [indice for indice in range(len(lugar['Género'])) if lugar['Género'][indice] == 'Masculino']
    subgrupo = BoW[Lugares_str[i]][indices]
    lsa_topic_matrix = lsa_model.fit_transform(subgrupo)
    for j, comp in enumerate(lsa_model.components_):
        terms_comp = zip(Caracteristicas[Lugares_str[i]], comp)
        sorted_terms = sorted(terms_comp, key=lambda x: x[1], reverse=True)[:10]
        print("Tópico " + str(j+1) + ": ")
        for t in sorted_terms:
            print(t[0], end=" - ")
        print("")
    i += 1
```

```python
i = 0
for lugar in Lugares:
    print("-"*90)
    print("Tópicos más relevantes en el subgrupo de mujeres - ", Lugares_str[i])
    lsa_model = TruncatedSVD(n_components=3)
    indices = [indice for indice in range(len(lugar['Género'])) if lugar['Género'][indice] == 'Femenino']
    subgrupo = BoW[Lugares_str[i]][indices]
    lsa_topic_matrix = lsa_model.fit_transform(subgrupo)
    for j, comp in enumerate(lsa_model.components_):
        terms_comp = zip(Caracteristicas[Lugares_str[i]], comp)
        sorted_terms = sorted(terms_comp, key=lambda x: x[1], reverse=True)[:10]
        print("Tópico " + str(j+1) + ": ")
        for t in sorted_terms:
            print(t[0], end=" - ")
        print("")
    i += 1
```

## 4. Palabras asociadas a opiniones negativas:

Ahora nos centramos sólo en las opiniones negativas para ver cuáles palabras tienen mayor relevancia:

```python
i = 0
vectorizer = CountVectorizer()
feats = SelectKBest(chi2, k=10)
for lugar in Lugares:
    print("Top 10 palabras más relevantes en opiniones negativas - ", Lugares_str[i])
    indices = [indice for indice in range(len(lugar['Escala'])) if lugar['Escala'][indice] < 3]
    opiniones_neg = lugar['Opinión'][indices]
    etiquetas = lugar['Escala'][indices]
    etiquetas = etiquetas.fillna(0)
    BoW_neg = vectorizer.fit_transform(opiniones_neg)
    Caracteristicas_neg = vectorizer.get_feature_names_out()
    feats.fit(BoW_neg, etiquetas)
    best = feats.get_support(indices=True)
    scores = feats.scores_
    sorted_scores_indices = sorted(enumerate(scores), key=lambda x: x[1], reverse=True)
    top_k_features = sorted_scores_indices[:10]
    w = {Caracteristicas_neg[k]: valor for k, valor in top_k_features}
    i += 1
    try:
        wordcloud = WordCloud(width=800, height=400).generate_from_frequencies(w)
        plt.imshow(wordcloud, interpolation='bilinear')
        plt.axis('off')
        plt.show()
    except:
        pass
```

Para este punto, como en el teatro Juárez y el Jardín Unión no se tienen opiniones negativas, se omite la nube de palabras en estos sitios.

## 5. Bolsas de Palabras de unigramas, bigramas y trigramas concatenadas:

Ya tenemos la bolsa de términos que creamos anteriormente, de ella se quieren obtener los términos que tienen el peso tfidf más alto para crear una bolsa de unigramas.

```python
i = 0
BoW_unig = dict()
Caracteristicas_unig = dict()
for lugar in Lugares:
    term_importance = np.sum(BoW[Lugares_str[i]].toarray(), axis=0)
    sorted_indices = np.argsort(term_importance)[::-1][:1000]
    top_terms = np.array(Caracteristicas[Lugares_str[i]])[sorted_indices]
    vectorizer_top = TfidfVectorizer(vocabulary=top_terms, norm='l2')
    BoW_unig[Lugares_str[i]] = vectorizer_top.fit_transform(lugar['Opinión'])
    Caracteristicas_unig[Lugares_str[i]] = vectorizer_top.get_feature_names_out()
    i += 1
```

Ahora, creamos las bolsas de bigramas:

```python
vectorizer_bigram = TfidfVectorizer(norm='l2', ngram_range=(2, 2))
i = 0
BoW_2 = dict()
Caracteristicas_2 = dict()
for lugar in Lugares:
    BoW_2[Lugares_str[i]] = vectorizer_bigram.fit_transform(lugar['Opinión'])
    Caracteristicas_2[Lugares_str[i]] = vectorizer_bigram.get_feature_names_out()
    i += 1
```

Igual que con los unigramas, vemos cuáles tienen un mayor peso tfidf para crear una bolsa de palabras con estos bigramas más importantes

```python
i = 0
BoW_big = dict()
Caracteristicas_big = dict()
for lugar in Lugares:
    term_importance = np.sum(BoW_2[Lugares_str[i]].toarray(), axis=0)
    sorted_indices = np.argsort(term_importance)[::-1][:2000]
    top_terms = np.array(Caracteristicas_2[Lugares_str[i]])[sorted_indices]
    vectorizer_top_bigramas = TfidfVectorizer(vocabulary=top_terms, norm='l2', ngram_range=(2, 2))
    BoW_big[Lugares_str[i]] = vectorizer_top_bigramas.fit_transform(lugar['Opinión'])
    Caracteristicas_big[Lugares_str[i]] = vectorizer_top_bigramas.get_feature_names_out()
    i += 1
```

Ahora viene la creación de trigramas:

```python
vectorizer_trigram = TfidfVectorizer(norm='l2', ngram_range=(3, 3))
i = 0
BoW_3 = dict()
Caracteristicas_3 = dict()
for lugar in Lugares:
    BoW_3[Lugares_str[i]] = vectorizer_trigram.fit_transform(lugar['Opinión'])
    Caracteristicas_3[Lugares_str[i]] = vectorizer_trigram.get_feature_names_out()
    i += 1
```

```python
i = 0
BoW_trig = dict()
Caracteristicas_trig = dict()
for lugar in Lugares:
    term_importance = np.sum(BoW_3[Lugares_str[i]].toarray(), axis=0)
    sorted_indices = np.argsort(term_importance)[::-1][:1000]
    top_terms = np.array(Caracteristicas_3[Lugares_str[i]])[sorted_indices]
    vectorizer_top_trigramas = TfidfVectorizer(vocabulary=top_terms, norm='l2', ngram_range=(3, 3))
    BoW_trig[Lugares_str[i]] = vectorizer_top_trigramas.fit_transform(lugar['Opinión'])
    Caracteristicas_trig[Lugares_str[i]] = vectorizer_top_trigramas.get_feature_names_out()
    i += 1
```

Y ahora concatenamos las bolsas de unigramas, bigramas y trigramas:

```python
from scipy.sparse import hstack
i = 0
BoW_concatenadas = dict()
Caracteristicas_concatenadas = dict()
for lugar in Lugares:
    BoW_concatenadas[Lugares_str[i]] = hstack([BoW_unig[Lugares_str[i]], BoW_big[Lugares_str[i]], BoW_trig[Lugares_str[i]]])
    Caracteristicas_concatenadas[Lugares_str[i]] = np.concatenate([Caracteristicas_unig[Lugares_str[i]], Caracteristicas_big[Lugares_str[i]], Caracteristicas_trig[Lugares_str[i]]])
    i += 1
```

Ahora, hacemos la selección de las mejores 1000 features y se muestran en una nube de palabras:

```python
i = 0
feats_1000 = SelectKBest(chi2, k=1000)
for lugar in Lugares:
    print("Top 1000 palabras más relevantes - ", Lugares_str[i])
    feats_1000.fit(BoW_concatenadas[Lugares_str[i]], lugar['Escala'])
    best = feats_1000.get_support(indices=True)
    scores = feats_1000.scores_
    sorted_scores_indices = sorted(enumerate(scores), key=lambda x: x[1], reverse=True)
    top_k_features = sorted_scores_indices[:1000]
    w = {Caracteristicas_concatenadas[Lugares_str[i]][k]: valor for k, valor in top_k_features}
    wordcloud = WordCloud(width=800, height=400).generate_from_frequencies(w)
    plt.imshow(wordcloud, interpolation='bilinear')
    plt.axis('off')
    plt.show()
    i += 1
```

## 6. Análisis temporal de las opiniones

```python
for lugar in Lugares:
    lugar['Fecha'] = pd.to_datetime(lugar['Fecha'], errors='coerce')
```

```python
i = 0
for lugar in Lugares:
    antiguos = lugar[lugar['Fecha'] < pd.Timestamp('2015-01-01')]['Escala']
    recientes = lugar[lugar['Fecha'] >= pd.Timestamp('2015-01-01')]['Escala']
    combined_series = pd.concat([antiguos, recientes])
    bins = np.histogram_bin_edges(combined_series.dropna(), bins=9) 
    plt.figure(figsize=(8, 5))
    plt.hist(antiguos.dropna(), bins=bins, label='Antiguos', align='mid', rwidth=0.5)
    plt.hist(recientes.dropna(), bins=bins, label='Recientes', align='right', rwidth=0.5)
    plt.title('Calificaciones a través del tiempo - '+ Lugares_str[i])
    plt.xlabel('Escala')
    plt.ylabel('Frecuencia')
    plt.legend(loc='upper left')
    plt.show()
    i += 1
```

De los histogramas podemos notar que en general el comportamiento de las calificaciones a través del tiempo es muy parecido, pues en todos los lugares la distribución de las calificaciones tiene la misma forma, excepto en el Callejón del beso, donde está la diferencia de que el máximo del histograma está en la calificación 4 para las opiniones antiguas, y para las recientes el máximo está en 5.

Además de estos histogramas, como herramienta para ver lo que estaba opinando la gente a través del tiempo podemos ver nubes de las palabras más relevantes antes de 2015 y después, lo cual se muestra a continuación:

```python
i = 0
vectorizer = CountVectorizer()
feats = SelectKBest(chi2, k=10)
for lugar in Lugares:
    print("Top 10 palabras más relevantes antes de 2015- ", Lugares_str[i])
    indices = [indice for indice in range(len(lugar['Escala'])) if lugar['Fecha'][indice] < pd.Timestamp('2015-01-01')]
    opiniones_2015 = lugar['Opinión'][indices]
    etiquetas = lugar['Escala'][indices]
    etiquetas = etiquetas.fillna(0)
    BoW_neg = vectorizer.fit_transform(opiniones_2015)
    Caracteristicas_neg = vectorizer.get_feature_names_out()
    feats.fit(BoW_neg, etiquetas)
    best = feats.get_support(indices=True)
    scores = feats.scores_
    sorted_scores_indices = sorted(enumerate(scores), key=lambda x: x[1], reverse=True)
    top_k_features = sorted_scores_indices[:10]
    w = {Caracteristicas_neg[k]: valor for k, valor in top_k_features}
    i += 1
    try:
        wordcloud = WordCloud(width=800, height=400).generate_from_frequencies(w)
        plt.imshow(wordcloud, interpolation='bilinear')
        plt.axis('off')
        plt.show()
    except:
        pass
```
